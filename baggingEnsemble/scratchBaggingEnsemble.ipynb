```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
```

```python
df=pd.DataFrame(load_iris().data, columns=load_iris().feature_names)
df['target']=pd.DataFrame(load_iris().target, columns=['target'])
```

```python
df.head()
```

```python
df=df[df['target']!=0][['sepal width (cm)','petal length (cm)','target']]
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt 
import seaborn as sns

sns.scatterplot(x='sepal width (cm)',y='petal length (cm)',hue='target',data=df)
```

```python
df=df.sample(100, replace=True)
df_train=df.iloc[:60,:]
df_test=df.iloc[60:,:]
```

```python
x_test=df_test.iloc[:,:-1]
y_test=df_test.iloc[:,-1]
```

# 1st model

```python
df_bag=df_train.sample(8,replace=True)
x=df_bag.iloc[:,:-1]
y=df_bag.iloc[:,-1]

df_test=df_test.sample(8,replace=True)
x_bag_test=df_test.iloc[:,:-1]
y_bag_test=df_test.iloc[:,-1]
```

```python
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.metrics import accuracy_score


bagging1=DecisionTreeClassifier()
estimate(bagging1,x,y,x_bag_test,y_bag_test)
```

```python
def estimate(clf,x,y, x_test,y_test):
    clf.fit(x,y)
    plot_tree(clf)
    y_pred=clf.predict(x_test)
    print(" accuracy :",accuracy_score(y_pred,y_test))
```
