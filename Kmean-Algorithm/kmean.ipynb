loading dataset

```python
import pandas as pd
import numpy as np

df=pd.read_csv("iq_cgpa_dataset.csv")
print(df.shape)
print(df.head())
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['CGPA'],df['IQ'])
```

```python
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,11):
    km=KMeans(n_clusters=i)
    km.fit_predict(df)
    wcss.append(km.inertia_)

print(wcss)
```

```python
plt.plot(range(1,11), wcss)
plt.xlabel("Number of Clusters")
plt.ylabel("wcss")
plt.show()
```

```python
X=df.iloc[:,:].values
km=KMeans(n_clusters=3)
y_means=km.fit_predict(X)
print(y_means)
```

```python
X[y_means==0,0]
```

```python
plt.scatter(X[y_means==0,1],X[y_means==0,0] , c="red")
plt.scatter(X[y_means==1,1],X[y_means==1,0] , c="blue")
plt.scatter(X[y_means==2,1],X[y_means==2,0] , c="green")
```
