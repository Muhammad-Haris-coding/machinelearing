```python
import numpy as np
import pickle
```

```python
ohe_sex=pickle.load(open('ohe_sex.pkl','rb'))
ohe_embarked=pickle.load(open('ohe_embarked.pkl','rb'))
clf1=pickle.load(open('decision_tree_model.pkl','rb'))
```

```python
test_data=np.array([1,'male',44.0,1,0,10.25,'S'],dtype=object).reshape(1,-1)
```

```python
test_data_sex=ohe_sex.transform(test_data[:,1].reshape(-1,1))
```

```python
test_data_embarked=ohe_embarked.transform(test_data[:,-1].reshape(-1,1))
```

```python
test_data_transformed=np.concatenate((test_data[:,[0,2,3,4,5]],test_data_sex,test_data_embarked),axis=1)
```

```python
clf1.predict(test_data_transformed)
```
