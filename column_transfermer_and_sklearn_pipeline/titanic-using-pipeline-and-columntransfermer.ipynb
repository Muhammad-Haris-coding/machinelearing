```python
import numpy as np
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import chi2
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline,make_pipeline
```

```python
df=pd.read_csv('titanic_data.csv')
df.head()
```

```python
df=df.drop(columns=['PassengerId','Name', 'Cabin','Ticket'])
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.drop(columns=['Survived']), df['Survived'],test_size=0.2, random_state=42)
```

```python
x_train.head()
```

# 1 Column Transfermer for the missing values in the age and embarked

```python
trf1=ColumnTransformer(transformers=[
    ('imputed_age',SimpleImputer(),[2]),
    ('imputed_embarked',SimpleImputer(strategy='most_frequent'),[6])
], remainder='passthrough')
```

# 2 onehot encoding of sex and Embarked column

```python
trf2=ColumnTransformer(transformers=[
    ('ohe_sex_embarked',OneHotEncoder(sparse_output=False,handle_unknown='ignore'),[1,6])
    ], remainder='passthrough')
```

# 3 scaling the feature through minmax scaling

```python
trf3=ColumnTransformer(transformers=[
    ('scale',MinMaxScaler(),slice(0,10))
    ], remainder='passthrough')
```

# 4 select kbest feature

```python
trf4=SelectKBest(score_func=chi2,k=8)
```

# 5 DecsionTreeClassifier

```python
trf5=DecisionTreeClassifier()
```

# 6 Now making pipelines

```python
pipe=Pipeline([
    ('trf1', trf1),
    ('trf2', trf2),
    ('trf3', trf3),
    ('trf4', trf4),
    ('trf5', trf5)
])
```

# 7 Fit the data

```python
pipe.fit(x_train,y_train)
```
