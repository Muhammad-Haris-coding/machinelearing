```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
import numpy as np
```

x,y=load_diabetes(return_X_y=True)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)

```python
from sklearn.linear_model import Ridge
```

```python
from sklearn.linear_model import SGDRegressor
```

# we can use SGD Regressor for gradient decsent ridge regression

```python
sgdr=SGDRegressor(alpha=0.01,learning_rate='constant',eta0=0.01 , max_iter=500, penalty='l2')
```

```python
sgdr.fit(x_train,y_train)
```

```python
y_pred=sgdr.predict(x_test)
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

```python
print(sgdr.coef_)
```

```python
print(sgdr.intercept_)
```

# we can also use the Ridge reggression but the gradient descent form instead the OLS

```python
y_pred=drge.predict(x_test)
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

# now i will make my own class for the gradient descent type of ridge regression

```python
class meraRidgeGD:
    def __init__ (self,learning_rate=0.01,epochs=100,alpha=0.1):
        self.lr=learning_rate
        self.epochs=epochs
        self.alpha=alpha
        self.coef_=None
        self.intercept_=None

    def fit(x_train,y_train):
        self.intercept_=0
        self.coef_=np.ones(x_train.shape[1])
        theta=np.insert(self.coef_,0,self.intercept_)
        x_train=np.insert(x_train,0,1,axis=1)

        for i in range(self.epochs):
            theta_dir=np.dot(x_train.T,x_train).dot(theta)-np.dot(x_train.T,y_train)+self.alpha*theta
            
            theta=theta+self.lr*theta_dir
            self.intercept_=theta[0]
            self.coef_=theta[1:]

    def predict(x_test):
        return np.dot(x_test,self.coef_)+self.intercept_
```
