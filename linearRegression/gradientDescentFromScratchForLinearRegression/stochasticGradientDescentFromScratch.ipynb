```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
x,y=load_diabetes(return_X_y=True)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
slr=LinearRegression()
```

```python
slr.fit(x_train,y_train)
```

```python
slr.intercept_
```

```python
slr.coef_
```

```python
y_pred=slr.predict(x_test)

from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

# Now i will make my own class from scratch for multiple column using stochastic Gradient descent

```python
class StochasticGradientDescent:
    def __init__ (self, learning_rate=0.1, epochs=100):
        self.lr=learning_rate
        self.epochs=epochs
        self.intercept_=None
        self.coef_=None

    def fit(self,x_train,y_train):
        self.intercept_=0
        self.coef_=np.ones(x_train.shape[1])
        for i in range (self.epochs):
            #another loop which will run according to the x_train rows size
            for j in range(x_train.shape[0]):
                idx=np.random.randint(0,x_train.shape[0])
                
                # this is for intercept_
                y_hat=np.dot(x_train[idx],self.coef_)+self.intercept_
                intercept_dirr=-2*(y_train[idx]-y_hat)
                self.intercept_=self.intercept_-(self.lr*intercept_dirr)
                
                #now for the coef_
                coef_dirr=-2*np.dot((y_train[idx]-y_hat),x_train[idx])
                self.coef_=self.coef_-(self.lr*coef_dirr)

    def predict(self,x_test):
        return np.dot(x_test,self.coef_)+self.intercept_
```

```python
sgd=StochasticGradientDescent(0.01,1000)
```

```python
sgd.fit(x_train,y_train)
```

```python
sgd.intercept_
```

```python
sgd.coef_
```

```python
y_pred=sgd.predict(x_test)

from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```
