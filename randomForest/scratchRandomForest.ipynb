```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_classification
```

```python
x,y=make_classification(n_features=5,n_redundant=0, n_informative=5 , n_clusters_per_class=1)
```

```python
df=pd.DataFrame(x,columns=['col1','col2','col3','col4','col5'])
df['target']=y
df.head()
```

```python
def sample_rows(df,percnt):
    return df.sample(int(df.shape[0]*percnt),replace=True)
```

```python
def sample_columns(df,percent):
    col=df.sample(df.columns.to_list()[:-1],int(df.shape[1]*percent))
    return df[col]
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
```

```python
df1=sample_rows(df,0.2)
```

```python
df2=sample_rows(df,0.2)
```

```python
df3=sample_rows(df,0.2)
```

```python
dt1=DecisionTreeClassifier()
dt2=DecisionTreeClassifier()
dt3=DecisionTreeClassifier()
```

```python
dt1.fit(df1.iloc[:,:-1],df1.iloc[:,-1])
dt2.fit(df2.iloc[:,:-1],df2.iloc[:,-1])
dt3.fit(df3.iloc[:,:-1],df3.iloc[:,-1])
```

```python
plot_tree(dt1)
```

```python
plot_tree(dt2)
```

```python
plot_tree(dt3)
```
